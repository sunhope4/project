## Install & Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

##Load Dataset Files

```python
customers = pd.read_csv("customers_dataset.csv")
sellers = pd.read_csv("sellers_dataset.csv")
orders = pd.read_csv(
    "orders_dataset.csv",
    engine="python",
    on_bad_lines="skip"
)
order_items = pd.read_csv(
    "order_items_dataset.csv",
    engine="python",
    on_bad_lines="skip"
)
payments = pd.read_csv(
    "order_payments_dataset.csv",
    engine="python",
    on_bad_lines="skip"
)
reviews = pd.read_csv(
    "order_reviews_dataset.csv",
    engine="python",
    on_bad_lines="skip"
)
products = pd.read_csv("products_dataset.csv")
```

##Merge Important Tables

```python
data = orders.merge(order_items, on="order_id")
data = data.merge(payments, on="order_id")
data = data.merge(reviews, on="order_id", how="left")
data = data.merge(products, on="product_id", how="left")
```

```python
# Convert time columns to datetime
data['order_purchase_timestamp'] = pd.to_datetime(data['order_purchase_timestamp'])
data['order_delivered_customer_date'] = pd.to_datetime(data['order_delivered_customer_date'])

# Create Month column for time analysis
data['order_month'] = data['order_purchase_timestamp'].dt.to_period('M')
```

##Monthly Sales Trend

```python
monthly_revenue = data.groupby('order_month')['price'].sum()

plt.figure(figsize=(10,5))
monthly_revenue.plot()
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

##Number of Orders per Month

```python
orders_month = data.groupby('order_month')['order_id'].nunique()

plt.figure(figsize=(10,5))
orders_month.plot()
plt.title("Orders per Month")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.show()
```

##Payment Type Distribution

```python
plt.figure(figsize=(8,6))
data['payment_type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Payment Method Distribution")
plt.ylabel("")
plt.show()
```

##Top 10 Product Categories

```python
top_categories = data['product_category_name'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_categories.values, y=top_categories.index)
plt.title("Top 10 Product Categories")
plt.xlabel("Number of Sales")
plt.ylabel("Category")
plt.show()
```

##Review Score Distribution

```python
plt.figure(figsize=(8,5))
sns.countplot(x='review_score', data=data)
plt.title("Review Score Distribution")
plt.show()
```

##Price vs Freight Cost

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='price', y='freight_value', data=data)
plt.title("Product Price vs Freight Cost")
plt.show()
```

##Seller Revenue Distribution

```python
seller_revenue = data.groupby('seller_id')['price'].sum()

plt.figure(figsize=(8,6))
sns.histplot(seller_revenue, bins=40)
plt.title("Seller Revenue Distribution")
plt.xlabel("Revenue")
plt.show()
```

##Delivery Time Analysis

```python
data['delivery_days'] = (
    data['order_delivered_customer_date'] - data['order_purchase_timestamp']
).dt.days

plt.figure(figsize=(8,6))
sns.histplot(data['delivery_days'].dropna(), bins=30)
plt.title("Delivery Time Distribution")
plt.xlabel("Days")
plt.show()
```

##Average Review Score per Category

```python
category_reviews = data.groupby('product_category_name')['review_score'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=category_reviews.values, y=category_reviews.index)
plt.title("Top Categories by Average Review Score")
plt.xlabel("Average Review Score")
plt.show()
```
